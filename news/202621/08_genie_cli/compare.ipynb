```python
from pathlib import Path
import shutil
import subprocess

before = Path("before.yml").read_text(encoding="utf-8")
after = Path("after.yml").read_text(encoding="utf-8")
```

```python
prompt = f"""Compare these bundle files. In two sentences, give the old and new job timeout
and explain whether deploying the bundle runs the job. Use only the supplied files.

before.yml
{before}

after.yml
{after}"""
```

```python
genie = shutil.which("genie.cmd") or shutil.which("genie")
if genie is None:
    raise FileNotFoundError("Install the Genie Code CLI and add it to PATH.")

result = subprocess.run(
    [genie, "exec", "--sandbox", "read-only", "--skip-git-repo-check", "--ephemeral", "-"],
    input=prompt, capture_output=True, text=True, encoding="utf-8",
)
if result.returncode:
    raise RuntimeError(result.stderr)
print(result.stdout.strip())
```
