```python
from pprint import pprint

sample = {
    "name": "accounts/sample-account/budget-spend-statuses/sample-budget",
    "alert_spend_statuses": [{
        "scope_type": "ALERT_CONFIGURATION_SCOPE_TYPE_PER_USER",
        "time_period": "MONTH",
        "spend": "12.50",
        "threshold": "50",
        "percent_consumed": "25.0",
    }],
}
print("Example monthly budget status")
pprint(sample)
```

```python
account_profile = ""
workspace_id = None
budget_id = ""
```

```python
if account_profile:
    from databricks.sdk import AccountClient

    account = AccountClient(profile=account_profile)
    parent = f"accounts/{account.config.account_id}"
    workspace_filter = f"workspace_id = {int(workspace_id)}" if workspace_id else None
    if budget_id:
        status = account.budgets.get_budget_spend_status(f"{parent}/budget-spend-statuses/{budget_id}")
        pprint(status.as_dict())
    else:
        page_token = None
        while True:
            page = account.budgets.list_budget_spend_statuses(parent, filter=workspace_filter, page_token=page_token)
            pprint([status.as_dict() for status in page.budget_spend_statuses or []])
            page_token = page.next_page_token
            if not page_token:
                break
```
